```python
import pandas as pd
df = pd.read_csv('orders_export.csv')

# a.
print("Missing counts:\n", df.isnull().sum())
print("\nMissing %:\n", (df.isnull().sum()/len(df))*100)

# b.
df['shipping_fee'] = df['shipping_fee'].fillna(df['shipping_fee'].median())
df['company'] = df['company'].fillna(df['company'].mode()[0])
print("\nAfter imputation:\n", df.isnull().sum())

# c.
before = len(df)
df = df.drop_duplicates()
after = len(df)
print(f"\nRows before: {before}, after: {after}")
```

```python
# TASK 8
Q1 = df['shipping_fee'].quantile(0.25)
Q3 = df['shipping_fee'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR
outliers_iqr = df[(df['shipping_fee'] < lower) | (df['shipping_fee'] > upper)]
print("IQR outliers:", len(outliers_iqr))

mean = df['shipping_fee'].mean()
std = df['shipping_fee'].std()
df['zscore'] = (df['shipping_fee'] - mean) / std
outliers_z = df[df['zscore'].abs() > 3]
print("Z-score outliers:", len(outliers_z))
```
